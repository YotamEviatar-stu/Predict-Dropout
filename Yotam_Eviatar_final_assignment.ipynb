## 1.4 One pass over every image

Per image: size, brightness, contrast, blur, saturation, and two pixel hashes.

```python
def laplacian_var(gray):
    g = gray.astype(np.float32)
    lap = (-4*g[1:-1,1:-1] + g[:-2,1:-1] + g[2:,1:-1] + g[1:-1,:-2] + g[1:-1,2:])
    return float(lap.var())

def ahash(g8):
    return "".join("1" if b else "0" for b in (g8 > g8.mean()).flatten())

rows, unreadable = [], []
for name in df["image_name"]:
    try:
        with open(os.path.join(IMG_DIR, name), "rb") as f:
            blob = f.read()                        # read once: md5 + decode from memory
        with Image.open(io.BytesIO(blob)) as im:
            w, h  = im.width, im.height
            rgb   = im.convert("RGB")
            small = np.asarray(rgb.resize((64, 64)), dtype=np.float32)
            g8    = np.asarray(rgb.convert("L").resize((8, 8)), dtype=np.float32)
    except Exception as e:
        unreadable.append((name, type(e).__name__)); continue

    gray = small.mean(axis=2)
    rows.append(dict(image_name=name, width=w, height=h, n_pixels=w*h,
                     brightness=gray.mean(), contrast=gray.std(),
                     blur=laplacian_var(gray),
                     saturation=(small.max(axis=2) - small.min(axis=2)).mean(),
                     md5=hashlib.md5(blob).hexdigest(), ahash=ahash(g8)))

print("images that could not be opened:", len(unreadable), unreadable[:3])
feat = pd.DataFrame(rows).merge(df, on="image_name", how="left")
feat["is_night"] = feat["saturation"] < 5          # IR captures are colourless

# Classes that will not be trained on: 'undefined' is not a species, and hyena (1 image) and
# wolf (9) cannot be evaluated at all - at 70/15/15 they get 0 and 1 test images.
# fox (47) is kept: dropping it would push field foxes into the nearest class, most likely
# jackal - one of the two target species.
DROP_CLASSES = ["undefined", "hyena", "wolf"]
feat  = feat[~feat["label"].isin(DROP_CLASSES)].reset_index(drop=True)
order = [c for c in counts.index if c not in DROP_CLASSES]
```

## 1.8 Day vs Night


```python
from scipy.stats import chi2_contingency

night = pd.crosstab(feat["label"], feat["is_night"])
night.columns = ["day", "night"]
night["night_%"] = (100*night["night"]/night.sum(axis=1)).round(1)
display(night.sort_values("night_%", ascending=False))

night[["day","night"]].div(night[["day","night"]].sum(axis=1), axis=0).plot(
    kind="barh", stacked=True, figsize=(7,4), color=["gold","navy"],
    title="Day / night share per class")
plt.tight_layout(); plt.show()

chi2, p, dof, _ = chi2_contingency(night[["day","night"]].values)
n = night[["day","night"]].values.sum()
V = np.sqrt(chi2 / (n * (min(night[["day","night"]].shape) - 1)))
print(f"label vs day/night: chi2={chi2:.1f}, p={p:.3g}, Cramer's V={V:.2f}")
```

## 1.9 Unusable images
I want to look for images that are too dark/bright/blurred and remove them later on

```python
from PIL import ImageOps

N = 10

def distinct(sub, n, min_bits=6):
    """Worst-first, keeping an image only if its 8x8 hash differs from those already kept, so one
    burst of near-identical frames cannot fill the row."""
    kept, seen = [], []
    for idx, h in zip(sub.index, sub["ahash"]):
        v = np.frombuffer(h.encode(), np.uint8)
        if all((v != q).sum() >= min_bits for q in seen):
            seen.append(v); kept.append(idx)
        if len(kept) == n:
            break
    return sub.loc[kept]

worst = [("too blurry",  distinct(feat.nsmallest(100, "blur"), N),       "blur"),
         ("too dark",    distinct(feat.nsmallest(100, "brightness"), N), "brightness"),
         ("too bright",  distinct(feat.nlargest(100, "brightness"), N),  "brightness")]
fig, axes = plt.subplots(3, N, figsize=(2*N, 7.5), squeeze=False)
for r, (name, sub, col) in enumerate(worst):
    for c in range(N):
        ax = axes[r, c]; ax.set_xticks([]); ax.set_yticks([])
        if c < len(sub):
            row = sub.iloc[c]
            im  = Image.open(os.path.join(IMG_DIR, row.image_name)).convert("RGB")
            ax.imshow(ImageOps.autocontrast(im))
            ax.set_title(f"{row.image_name.split('_')[-1].split('.')[0]}\n{row.label} | {col} {row[col]:.0f}",
                         fontsize=6)
    axes[r, 0].set_ylabel(name, fontsize=9)
fig.suptitle(f"worst {N} per category, after autocontrast")
plt.tight_layout(); plt.show()

for name, sub, col in worst:
    print(f"\n{name}:")
    for _, row in sub.iterrows():
        iid = row.image_name.split("_")[-1].split(".")[0]
        print(f'    "{iid}",   # {row.label}, {col} {row[col]:.0f}')
```

## 1.10 Duplicates and burst frames

Check for duplicates

```python
exact = [g for g in feat.groupby("md5")["image_name"].apply(list) if len(g) > 1]
print("byte-identical groups:", len(exact), "| redundant copies:", sum(len(g)-1 for g in exact))

img_names = feat["image_name"].tolist()
mat   = np.stack([np.frombuffer(h.encode(), dtype=np.uint8) for h in feat["ahash"]])
pairs = []
for i in range(len(img_names)):
    d = (mat[i+1:] != mat[i]).sum(axis=1)
    for j in np.where(d <= 5)[0]:
        pairs.append((img_names[i], img_names[i+1+j], int(d[j])))
print("near-duplicate pairs (hamming <= 5):", len(pairs))

lab   = df.set_index("image_name")["label"].to_dict()
cross = [(a,b,d) for a,b,d in pairs if lab.get(a) != lab.get(b)]
print("near-duplicate pairs with DIFFERENT labels (possible label noise):", len(cross))
for a,b,d in cross[:5]:
    print(f"  {a} [{lab.get(a)}] vs {b} [{lab.get(b)}] d={d}")

# near-duplicate pairs per class: how many, and the 3 tightest examples of each
pdf = pd.DataFrame(pairs, columns=["a", "b", "d"])
pdf["label_a"], pdf["label_b"] = pdf["a"].map(lab), pdf["b"].map(lab)
same = pdf[(pdf.label_a == pdf.label_b) & pdf.label_a.isin(order)]
print("\nnear-duplicate pairs within a class:")
display(same["label_a"].value_counts().reindex(order).fillna(0).astype(int).to_frame("pairs"))

for cls in order:
    top = same[same.label_a == cls].nsmallest(3, "d")
    if not len(top):
        print(f"{cls}: no near-duplicate pairs"); continue
    fig, axes = plt.subplots(1, 2*len(top), figsize=(2.2*2*len(top), 2.6))
    axes = np.atleast_1d(axes)
    for k, (_, r) in enumerate(top.iterrows()):
        for off, nme in enumerate([r.a, r.b]):
            ax = axes[2*k + off]
            ax.imshow(Image.open(os.path.join(IMG_DIR, nme)).convert("RGB"))
            ax.set_title(f"d={r.d}\n{nme[-12:]}", fontsize=7); ax.axis("off")
    fig.suptitle(f"{cls} - 3 tightest near-duplicate pairs", fontsize=10)
    plt.tight_layout(); plt.show()
```

1. No exact duplicates found.

## 2.3 Train / validation / test split — 70 / 15 / 15

Split before any image is touched; test is opened once, at the end. `new_dataset/` is Task-5 input, not a test set.

Single hold-out, no K-fold.

```python
SEED, FRAC = 42, {"train": 0.70, "val": 0.15, "test": 0.15}
SPLITS     = ["train", "val", "test"]

lab    = clean.set_index("image_name")["label"].to_dict()
parent = {n: n for n in clean["image_name"]}
def find(x):
    while parent[x] != x:
        parent[x] = parent[parent[x]]; x = parent[x]
    return x

kept_pairs = [(a, b) for a, b, _ in pairs if a in parent and b in parent]
same_label = [(a, b) for a, b in kept_pairs if lab[a] == lab[b]]
for a, b in same_label:                                  # cross-label pairs are hash collisions, not bursts
    ra, rb = find(a), find(b)
    if ra != rb: parent[ra] = rb

clean["group"] = [find(n) for n in clean["image_name"]]
g = (clean.groupby("group")
          .agg(label=("label", "first"), n=("image_name", "size"), nlab=("label", "nunique"))
          .reset_index())
assert g["nlab"].max() == 1, "a group carries two species"

def quota(n):                                            # largest remainder, so the three parts sum to n
    raw = {s: n*f for s, f in FRAC.items()}
    q   = {s: int(np.floor(v)) for s, v in raw.items()}
    for s in sorted(FRAC, key=lambda s: (-(raw[s]-q[s]), SPLITS.index(s)))[:n - sum(q.values())]:
        q[s] += 1
    return q

assign = {}
for c in CLASSES:
    gc = (g[g.label == c].sample(frac=1, random_state=SEED)          # reproducible tie-break between
                         .sort_values("n", kind="stable", ascending=False))   # equal-sized groups
    q, filled = quota(int(gc.n.sum())), {s: 0 for s in SPLITS}
    for r in gc.itertuples():                                        # largest group -> emptiest split
        s = max(SPLITS, key=lambda s: (q[s] - filled[s], q[s]))
        assign[r.group] = s; filled[s] += r.n
clean["split"] = clean["group"].map(assign)

tab = pd.crosstab(clean["label"], clean["split"])[SPLITS].reindex(CLASSES)
tab["n"] = tab.sum(axis=1)
for s in SPLITS: tab[s + "_%"] = (100*tab[s]/tab["n"]).round(1)
display(tab)

for a, b in [("train", "val"), ("train", "test"), ("val", "test")]:
    assert not (set(clean.loc[clean.split == a, "group"]) & set(clean.loc[clean.split == b, "group"]))
cross = len(kept_pairs) - len(same_label)
print(f"{len(clean)} images in {len(g)} groups | {len(same_label)} same-label pairs merged | "
      f"{cross} cross-label pairs left unmerged | {len(pairs)-len(kept_pairs)} pairs touch a dropped image")
print("no group crosses a split boundary | " + " | ".join(f"{s} {int(tab[s].sum())}" for s in SPLITS))
print(f"smallest test class {tab['test'].idxmin()}={int(tab['test'].min())} -> its per-class recall "
      f"can only move in steps of {100/tab['test'].min():.1f} pp")
```

2. At that size one test image is 14.3 points of fox recall and 11.1 for fallow_deer / porcupine.

Augmentation, train split only, Ultralytics defaults:

| transform | value | why it fits camera-trap crops |
|---|---|---|

| `RandomHorizontalFlip` | p = 0.5 | facing left is the same species as facing right |
| `RandomVerticalFlip` | off | a camera trap is never upside down |
| `RandAugment` | default | colour and brightness variation, so "grey = nocturnal" is not free |
| `RandomErasing` | p = 0.4 | forces use of more than one body part |
| mosaic / mixup | off | detection-only |

# Step 4 — Hyperparameter tuning

## 4.1 Parameters, values and hypotheses

Baseline from Step 3: `batch=16`, `imgsz=224`, `weight_decay=0.0005`. Each parameter is moved one step down and one step up, everything else fixed (`seed=42` included) — 6 extra runs, one-at-a-time. Values: `batch` 8 / **16** / 32 (powers of two), `imgsz` 160 / **224** / 320 (multiples of 32), `weight_decay` 0.0 / **0.0005** / 0.005 (log scale).

- `batch=8`: noisier gradients, less stable and slower per epoch, choppier curves.
- `batch=32`: smoother, faster updates; less gradient noise, so macro recall on small classes may fall.
- `imgsz=160`: faster, but small animals (porcupine, fox) lose detail, so lower recall on those.
- `imgsz=320`: more pixels should separate similar species — higher macro recall at more compute.
- `weight_decay=0.0`: no penalty, so the model fits training data more freely and overfits.
- `weight_decay=0.005`: 10x stronger penalty cuts overfitting on the rare classes; too strong underfits.

```python

BASE = dict(data=DATA_ROOT, project=OUTPUT_DIR / "runs", imgsz=224, epochs=30,
            seed=42, deterministic=True, device=DEVICE)

RUNS = [
    ("batch_8",    dict(batch=8)),
    ("batch_32",   dict(batch=32)),
    ("imgsz_160",  dict(imgsz=160)),
    ("imgsz_320",  dict(imgsz=320)),
    ("wd_0.0",     dict(weight_decay=0.0)),
    ("wd_0.005",   dict(weight_decay=0.005)),
]

for name, override in RUNS:
    if (OUTPUT_DIR / "runs" / name / "weights" / "best.pt").exists():
        print("skip (already trained):", name); continue
    YOLO("yolo26n-cls.pt").train(**{**BASE, **override}, name=name, exist_ok=True)
    print("done:", name)
```

## 4.2 Validate all seven

Validation only — scoring the sweep on test would turn it into a second validation set. Ranking = mean of macro P, macro R and top-5; top-5 is ~1.0 everywhere, so P and R decide it.

```python

def val_scores(run_dir):
    m = YOLO(run_dir / "weights" / "best.pt").val(data=DATA_ROOT, split="val",
                                                  device=DEVICE, verbose=False)
    M = m.confusion_matrix.matrix.T
    P = np.diag(M) / np.maximum(M.sum(0), 1)
    R = np.diag(M) / np.maximum(M.sum(1), 1)
    gz, jk = names.index("gazelle"), names.index("jackal")
    return dict(macroP=P.mean(), macroR=R.mean(), top1=m.top1, top5=m.top5,
                R_gazelle=R[gz], R_jackal=R[jk],              # the two target species, reported not selected on
                gz_jk_swaps=int(M[gz, jk] + M[jk, gz]))       # must stay 0: the confusion that would matter most

rows = {"baseline": val_scores(RUN)}
for name, _ in RUNS:
    rows[name] = val_scores(OUTPUT_DIR / "runs" / name)

comp = pd.DataFrame(rows).T
comp["score"] = comp[["macroP", "macroR", "top5"]].mean(axis=1)
comp = comp.sort_values("score", ascending=False)
winner = comp.index[0]
print(comp.round(3))
print("\nwinner:", winner)

comp.round(4).to_csv(OUTPUT_DIR / "task4_val_comparison.csv")
```

Validation, score = mean(macro P, macro R, top-5):

| run | macro P | macro R | top-1 | score |
|---|---|---|---|---|
| **batch_8** | **0.906** | **0.892** | **0.950** | **0.932** |
| imgsz_160 | 0.897 | 0.870 | 0.944 | 0.922 |
| batch_32 | 0.894 | 0.863 | 0.938 | 0.919 |
| wd_0.005 | 0.878 | 0.873 | 0.938 | 0.916 |

```python
tuned_dir = RUN if winner == "baseline" else OUTPUT_DIR / "runs" / winner
tuned = YOLO(tuned_dir / "weights" / "best.pt")
names = [tuned.names[i] for i in range(len(tuned.names))]

m = tuned.val(data=DATA_ROOT, split="test", device=DEVICE, verbose=False)
M = m.confusion_matrix.matrix.T
P = np.diag(M) / np.maximum(M.sum(0), 1)
R = np.diag(M) / np.maximum(M.sum(1), 1)

print(f"tuned model = {winner}")
print(f"test: macro P {P.mean():.3f} | macro R {R.mean():.3f} | top-1 {m.top1:.3f} | top-5 {m.top5:.3f}")
print(pd.DataFrame({"test_n": M.sum(1).astype(int), "P": P.round(3), "R": R.round(3)}, index=names))

g, j, d = (names.index(c) for c in ("gazelle", "jackal", "fallow_deer"))
print(f"\ngazelle R {R[g]:.3f}   fallow_deer R {R[d]:.3f}   jackal R {R[j]:.3f}")
print(f"gazelle<->jackal      : {int(M[g, j])} / {int(M[j, g])}")
print(f"gazelle<->fallow_deer : {int(M[g, d])} / {int(M[d, g])}")

pd.DataFrame({"test_n": M.sum(1).astype(int), "P": P.round(3), "R": R.round(3)},
             index=names).to_csv(OUTPUT_DIR / "task4_test_per_class.csv")
pd.DataFrame(M.astype(int), index=names, columns=names).to_csv(OUTPUT_DIR / "task4_test_confusion.csv")
```

1. The tuned model did not beat the baseline on test — 0.9 points of macro recall is 2 images of 320, after leading by 4.3 points on validation. The pick stands: it was made on validation.
2. They disagree because with 7-9 images per rare class macro recall moves in 11-14 point jumps — the cost of one hold-out instead of K-fold.
3. On test, gazelle and jackal are never confused. Gazelle's only look-alike error is fallow_deer, 1 each way (P = 0.95, R = 0.86). Jackal P = 1.00 (no fox or other species became jackal); its errors go to wild boar (3) and fox (1).


## 4.5 Rare-class oversampling

Fox, fallow deer and porcupine have few training images (33 / 41 / 44), so their training images are copied (x4 / x3 / x3) to about 130 each, with an oversampling technique. Train split only; val and test are unchanged. Only with `batch_8` model

```python
import shutil

REPEAT  = {"fox": 4, "fallow_deer": 3, "porcupine": 3}      # duplication factor, train only
OS_ROOT = WORK_DIR / "data_cls_os"
OS_RUN  = OUTPUT_DIR / "runs" / "oversample"

if not OS_ROOT.exists():
    shutil.copytree(DATA_ROOT, OS_ROOT)                     # val/ and test/ carried over as-is
    for c, k in REPEAT.items():
        for f in sorted((OS_ROOT / "train" / c).glob("*")):
            for i in range(2, k + 1):
                shutil.copy(f, f.with_name(f"{f.stem}_dup{i}{f.suffix}"))

print({n: len(list((OS_ROOT / "train" / n).glob("*"))) for n in names})

best_cfg = {} if winner == "baseline" else dict(RUNS)[winner]
if (OS_RUN / "weights" / "best.pt").exists():
    print("skip (already trained):", OS_RUN)
else:
    YOLO("yolo26n-cls.pt").train(**{**BASE, **best_cfg, "data": OS_ROOT},
                                 name="oversample", exist_ok=True)
```

```python
gz, fd = (names.index(c) for c in ["gazelle", "fallow_deer"])
def rare_counts(run_dir):                                   # correct / total, val, rare classes
    m = YOLO(run_dir / "weights" / "best.pt").val(data=DATA_ROOT, split="val",
                                                  device=DEVICE, verbose=False)
    M = m.confusion_matrix.matrix.T
    return {n: f"{int(M[i, i])}/{int(M[i].sum())}" for i, n in enumerate(names) if n in REPEAT}

def pair_check(run_dir):                                    # gazelle <-> fallow_deer on val
    m  = YOLO(run_dir / "weights" / "best.pt")
    pr = {c: np.stack([r.probs.data.cpu().numpy() for r in
                       m.predict([str(f) for f in sorted((DATA_ROOT / "val" / c).glob("*"))],
                                 device=DEVICE, verbose=False)])
          for c in ["gazelle", "fallow_deer"]}
    swaps = int((pr["gazelle"].argmax(1) == fd).sum() + (pr["fallow_deer"].argmax(1) == gz).sum())
    return f"{swaps}", round(float(pr["fallow_deer"][:, gz].mean()), 3)

comp_os = pd.DataFrame({winner: val_scores(tuned_dir), "oversample": val_scores(OS_RUN)}).T
for r in comp_os.index:
    rd = tuned_dir if r == winner else OS_RUN
    for k, v in rare_counts(rd).items():
        comp_os.loc[r, k] = v
    comp_os.loc[r, "gz<->fd swaps"], comp_os.loc[r, "p(gazelle | true fallow_deer)"] = pair_check(rd)
display(comp_os.round(3))
```

1. Better on validation: macro P 0.906 -> 0.945, macro R 0.892 -> 0.916, top-1 0.950 -> 0.960, mostly from fallow deer (5/9 -> 8/9) and gazelle <-> `fallow_deer` swaps 3 -> 0. The cost: fox 6/7 -> 5/7 and gazelle recall 0.913 -> 0.870.
2. Not used for Step 5: the brief selects the final model from the six hyperparameter runs, and oversampling changes the data, not a hyperparameter. It is reported as a possible improvement; Step 5 uses `batch_8`.

## 5.2 Crop generation

```python
from PIL import Image
import shutil

if REUSE:                                            # crops already written by an earlier run
    field = pd.DataFrame({"crop": sorted(f.name for f in CROP_DIR.glob("*.JPG"))})
    field["frame"] = field["crop"].str.rsplit("_", n=1).str[0] + ".JPG"
else:
    if CROP_DIR.exists():
        shutil.rmtree(CROP_DIR)
    CROP_DIR.mkdir(parents=True)

    rows = []
    for r in ok:
        im = None
        for k, d in enumerate(r["detections"]):
            if d["category"] != "1" or d["conf"] < MD_CONF:      # animals only, above the cutoff
                continue
            if im is None:
                im = Image.open(r["file"]).convert("RGB")
            W, H = im.size
            x, y, w, h = d["bbox"]
            fname = Path(r["file"]).stem + f"_{k}.JPG"
            im.crop((int(x * W), int(y * H), int((x + w) * W), int((y + h) * H))).save(CROP_DIR / fname, quality=90)
            rows.append({"crop": fname, "frame": Path(r["file"]).name, "det_conf": round(d["conf"], 3)})
    field = pd.DataFrame(rows)

print(f"{len(field)} crops from {field['frame'].nunique()} of {n_frames} frames -> {CROP_DIR}")
```

## 5.3 Species classification

## 5.4 Rejecting crops the model should not label

MegaDetector returns crops no class fits: animals outside the label set (`hyena`, `wolf`) and boxes with no identifiable animal (the `undefined` case). Softmax cannot say "none of these", so both come out as a confident species.

'tau' will represent threshold

Reject rule: top-1 below `tau` → `uncertain`, with `tau` set on validation as the 5th percentile of top-1 confidence among correct crops. Tested on the `undefined`, `hyena` and `wolf` crops, which never entered any split.

```python
# threshold comes from validation only - the field set has no labels to tune against
vfiles = [(f, c) for c in names for f in sorted((DATA_ROOT / "val" / c).glob("*"))]
vp = np.stack([r.probs.data.cpu().numpy() for r in                   # stream: one result at a time
               clf.predict([str(f) for f, _ in vfiles], device=DEVICE, stream=True, verbose=False)])
v_conf = vp.max(1)
v_ok   = vp.argmax(1) == np.array([names.index(c) for _, c in vfiles])
TAU    = float(np.percentile(v_conf[v_ok], 5))       # keep 95% of the calls already correct

rej = v_conf < TAU
print(f"tau = {TAU:.3f}")
print(f"val : {rej.sum()} of {len(vfiles)} rejected - {int((~v_ok & rej).sum())} were wrong, "
      f"{int((v_ok & rej).sum())} were right ({int((~v_ok).sum())} errors in total)")
print(f"      top-1 {v_ok.mean():.3f} over all {len(vfiles)} -> {v_ok[~rej].mean():.3f} over the {int((~rej).sum())} kept")

# out-of-class check: these classes were dropped in Step 1, so no split ever contained them
ooc = df[df["label"].isin(DROP_CLASSES)].copy()
op  = np.stack([r.probs.data.cpu().numpy() for r in
                clf.predict([os.path.join(IMG_DIR, n) for n in ooc["image_name"]],
                            device=DEVICE, stream=True, verbose=False)])
ooc["pred"], ooc["conf"] = [names[i] for i in op.argmax(1)], op.max(1)
ooc["rejected"] = ooc["conf"] < TAU

print("\nnever-trained classes, at this tau:")
display(ooc.groupby("label").agg(n=("conf", "size"), rejected=("rejected", "sum"),
                                 mean_conf=("conf", "mean")).round(3))
display(ooc[ooc["label"] != "undefined"][["label", "pred", "conf", "rejected"]]
        .sort_values("conf", ascending=False).round(3))
print("undefined crops that survive, called:",
      ooc[(ooc.label == "undefined") & ~ooc.rejected]["pred"].value_counts().head(4).to_dict())

field["species_final"] = np.where(field["conf"] < TAU, "uncertain", field["species"])
print(f"\nfield: {int((field['conf'] < TAU).sum())} of {len(field)} crops rejected "
      f"({100*(field['conf'] < TAU).mean():.1f}%, against {100*rej.mean():.1f}% on validation)")
```

1. `tau` = 0.82 rejects 22 of 321 validation crops (6 wrong, 16 right); top-1 over survivors 0.950 -> 0.967.
2. Out-of-class animals are caught: 8 of the 10 hyena and wolf crops rejected, including the hyena called `jackal` at 0.642.
3. In the field 204 of 993 crops (20.5%) are rejected against 6.9% on validation — the distribution shift, since MegaDetector keeps partial, distant and empty boxes.

## 5.5 Distribution and reliability

`count_x_P` is each count scaled by that class's test precision.

```python
order_f = names + ["uncertain"]
counts  = field["species_final"].value_counts().reindex(order_f).fillna(0).astype(int)

fig, ax = plt.subplots(figsize=(8, 4))
TARGETS = ["gazelle", "jackal"]                   # the two species the study is about
bar_c   = ["darkorange" if c in TARGETS else "steelblue" for c in names] + ["darkgrey"]
counts.plot(kind="bar", ax=ax, color=bar_c)
ax.set_ylabel("crops"); ax.set_title(f"May 2022 field data - species counts after rejection (tau = {TAU:.2f})")
for i, v in enumerate(counts):
    ax.text(i, v, str(v), ha="center", va="bottom", fontsize=8)
plt.tight_layout(); plt.show()

rel = pd.DataFrame({"raw": field["species"].value_counts().reindex(names).fillna(0).astype(int),
                    "kept": counts.reindex(names),
                    "test_P": pd.Series(P, index=names).round(2),
                    "test_R": pd.Series(R, index=names).round(2)})
rel["kept_x_P"] = (rel["kept"] * rel["test_P"]).round().astype(int)
print(rel)
print(f"\n{int(counts['uncertain'])} of {len(field)} crops rejected at tau = {TAU:.3f}")

field.to_csv(OUTPUT_DIR / "field_predictions.csv", index=False)
rel.to_csv(OUTPUT_DIR / "field_species_counts.csv")
```

3. Scaling by precision barely moves any count (457 -> 425 wild boar, 13 -> 12 gazelle). The 204 rejected crops are the weak link, not precision.

## 5.6 Reliability — conclusion

**Inference chain:** 2,423 frames → MegaDetector (conf >= 0.2, animals only) → 993 crops from 710 frames → `batch_8` classifier, top-1 label → `uncertain` if top-1 < tau = 0.82 (204 crops).

**How the model separates gazelle, fallow deer and jackal** — `batch_8` on the test set (rows = true):

| true \ predicted | gazelle | fallow_deer | jackal | other |
|---|---|---|---|---|
| gazelle (22) | **19** | 1 | 0 | 2 (birds 1, cow 1) |
| fallow_deer (9) | 1 | **7** | 0 | 1 (wild_boar) |
| jackal (33) | 0 | 0 | **29** | 4 (fox 1, wild_boar 3) |

1. **Gazelle and jackal are never confused:** 0 swaps in either direction, on test and on validation. So the field result (jackal 18 vs gazelle 13 kept, 37 vs 22 raw) is not caused by the model mixing the two.
2. **Gazelle and fallow deer are the only real confusion:** 1 swap each way on test (gazelle R = 0.86, P = 0.95). Fallow deer are rare in the field (4 crops), so at most about 1 crop wrongly becomes gazelle.
3. **Jackal is never over-counted:** no other species was called jackal on test (P = 1.00), including 0 of 7 foxes. Its errors go the other way (3 jackals called wild boar, 1 called fox; R = 0.88), so the field jackal count can only be slightly low.
4. **The model is less sure on field data than on test:** the target species have the least certain predictions (jackal mean confidence 0.77, 19 of 37 crops rejected; gazelle 9 of 22), compared with 83 of 540 for wild boar. Field crops include partial and distant animals, and 260 of the 344 `undefined` crops still pass tau, so an out-of-class crop can still be counted as any species.
5. **Bottom line:** on test the model separates gazelle, jackal and fallow deer well (precision 0.95 / 1.00 / 0.88, recall 0.86 / 0.88 / 0.78) and never confuses gazelle with jackal. On field data it is less sure (mean confidence 0.84 / 0.77 / 0.84, about half of target crops rejected), so the counts show that jackals outnumber gazelles, not how many individuals there are.
